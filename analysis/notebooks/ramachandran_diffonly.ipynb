```python
import matplotlib.pyplot as plt

from analysis.topology import get_topology_from_mol
from analysis.dihedrals import get_dihedrals
```

```python
mol_path = "/Users/marl/Code/STDDPM_v2/analysis/data/ala2.mol"
equi_path = "/Users/marl/Code/STDDPM_v2/analysis/data/equi_DDPM"
nonequi_path = "/Users/marl/Code/STDDPM_v2/analysis/data/non_equi_DDPM"

# Get topology
top = get_topology_from_mol(mol_path)

# Get dihedrals
phi_atoms = [0, 8, 6, 1]
psi_atoms = [8, 6, 1, 3]
dihedrals = {}
for key, coord_file_path in zip(["equi", "nonequi"], [equi_path, nonequi_path]):
    dihedrals[key] = {}
    dihedrals[key]["phi"], dihedrals[key]["psi"] = get_dihedrals(coord_file_path, top, phi_atoms, psi_atoms, save_pdb=True)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.scatter(dihedrals["equi"]["phi"], dihedrals["equi"]["psi"], s=10)
plt.xlabel(r"$\phi$", fontsize="x-large", labelpad=10)
plt.ylabel(r"$\psi$", fontsize="x-large", labelpad=10)
plt.title("Equi")
plt.subplot(122)
plt.scatter(dihedrals["nonequi"]["phi"], dihedrals["nonequi"]["psi"], s=10)
plt.xlabel(r"$\phi$", fontsize="x-large", labelpad=10)
plt.ylabel("\n"+r"$\psi$", fontsize="x-large", labelpad=10)
plt.title("Non-Equi")

plt.tight_layout()
plt.show()
```
