```python


# Get topology
top = get_topology_from_mol(mol_path)
print([a for a in top.atoms])
# Get dihedrals
phi_atoms = [0, 8, 6, 1]
psi_atoms = [8, 6, 1, 3]
dihedrals = {}
for key, coord_file_path in zip(["20", "800", "900"], [path_20, path_800, path_900]):
    dihedrals[key] = {}
    dihedrals[key]["phi"], dihedrals[key]["psi"] = get_dihedrals(coord_file_path, top, phi_atoms, psi_atoms, save_pdb=True, file_type="mol")
```

```python
plt.figure(figsize=(15, 5))

plt.subplot(131)
plt.scatter(dihedrals["20"]["phi"], dihedrals["20"]["psi"], s=10)
plt.xlabel(r"$\phi$", fontsize="x-large", labelpad=10)
plt.ylabel(r"$\psi$", fontsize="x-large", labelpad=10)
plt.title("Dynamic_20")
plt.subplot(132)
plt.scatter(dihedrals["800"]["phi"], dihedrals["800"]["psi"], s=10)
plt.xlabel(r"$\phi$", fontsize="x-large", labelpad=10)
plt.ylabel("\n"+r"$\psi$", fontsize="x-large", labelpad=10)
plt.title("Dynamic_800")
plt.subplot(133)
plt.scatter(dihedrals["900"]["phi"], dihedrals["900"]["psi"], s=10)
plt.xlabel(r"$\phi$", fontsize="x-large", labelpad=10)
plt.ylabel("\n"+r"$\psi$", fontsize="x-large", labelpad=10)
plt.title("Dynamic_900")

plt.tight_layout()
plt.show()
```
